```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)
df['age'] = df['age'].fillna(df['age'].mean())
df = df.dropna(subset=['embarked'])
df.head()
```

```python
#encode categorical columns

# Simple label encoding for sex
df['sex_encoded'] = df['sex'].map({'male': 0, 'female': 1})

# One-hot encoding for embarked
df = pd.get_dummies(df, columns=['embarked'], prefix='embarked')

df.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
df[['age', 'fare']] = scaler.fit_transform(df[['age', 'fare']])
df[['age', 'fare']].head()
```

Categorical columns like `sex` and `embarked` were converted into numeric form so they can be used in machine learning models. Numeric columns like `age` and `fare` were scaled so they have a similar range (mean 0, standard deviation 1), which helps many models perform better
