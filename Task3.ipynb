```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)
df.head()
```

```python
print(df.shape)
df.info()
```

```python
print(df.isnull().sum())
```

```python
#fill missing age with the average age
df['age'] = df['age'].fillna(df['age'].mean())

#drop rows where 'embarked' is missing (only a few)
df = df.dropna(subset=['embarked'])
print(df.isnull().sum())
```
