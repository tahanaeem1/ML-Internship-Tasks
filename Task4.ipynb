```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
#Bar Chart
df['class'].value_counts().plot(kind='bar')
plt.title("Passenger Count by Class")
plt.xlabel("Class")
plt.ylabel("Number of Passengers")
plt.show()
```

Most passengers were in Third Class

```python
#Line Chart
avg_fare_by_class = df.groupby('pclass')['fare'].mean()
avg_fare_by_class.plot(kind='line', marker='o')
plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.show()
```

Average fare shows a sharp drop from 1st to third

```python
#histogram
df['age'].plot(kind='hist', bins=20)
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

Most passengers were between 20 to 40 year olds
